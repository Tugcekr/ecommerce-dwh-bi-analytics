```python
import pandas as pd
from sqlalchemy import create_engine
import os
import urllib

# 1. Klasör yolun
klasor_yolu = r"C:\Users\Tugce\OneDrive\Masaüstü\Proje_Omnichannel_DW" 

# 2. SQL Server'a Modern Köprü Kuruyoruz (Hata çözen kısım burası!)
params = urllib.parse.quote_plus(
    r'Driver={ODBC Driver 17 for SQL Server};'
    r'Server=localhost\SQLEXPRESS;'
    r'Database=ECommerce_DW;'
    r'Trusted_Connection=yes;'
)
engine = create_engine(f"mssql+pyodbc:///?odbc_connect={params}", fast_executemany=True)

# 3. Yüklenecek Dosyalar ve SQL'deki Kurumsal İsimleri
dosyalar_ve_tablolar = {
    "monthly_budget_targets.csv": "stg_monthly_budget",
    "order_vouchers_shipping.csv": "stg_order_vouchers",
    "distribution_centers.csv": "stg_distribution_centers",
    "products_old.csv": "stg_products",
    "users_old.csv": "stg_users",
    "orders.csv": "stg_orders",
    "order_items.csv": "stg_order_items",
    "inventory_items_old.csv": "stg_inventory_items",
    "events_old.csv": "stg_events"
}

print("Veriler SQL'e yüklenmeye başlıyor. fast_executemany devrede...\n")

for dosya_adi, tablo_adi in dosyalar_ve_tablolar.items():
    tam_yol = os.path.join(klasor_yolu, dosya_adi)
    
    if os.path.exists(tam_yol):
        print(f"⏳ {dosya_adi} okunuyor ve SQL'e '{tablo_adi}' olarak yükleniyor...")
        
        # Veriyi okuyoruz
        df = pd.read_csv(tam_yol)
        
        # SQL'e fırlatıyoruz!
        df.to_sql(tablo_adi, con=engine, if_exists='replace', index=False, chunksize=50000)
        
        print(f"  ✅ Yüklendi! ({len(df)} satır)")
    else:
        print(f"  ❌ DİKKAT: {dosya_adi} bulunamadı!")

print("\n🎉 TEBRİKLER! Tüm ETL (Veri Aktarımı) işlemi başarıyla tamamlandı.")
```
